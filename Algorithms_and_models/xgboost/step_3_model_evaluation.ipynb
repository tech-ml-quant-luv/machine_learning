```python
import numpy as np
import pandas as pd
from sklearn.model_selection import KFold, cross_val_score, StratifiedKFold
from xgboost import XGBClassifier, plot_tree
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme()
```

    Evaluate Models with K-Fold Cross-Validation

```python
df = pd.read_csv("pima-indians-diabetes.csv")
X = df.iloc[:,0:8]
y = df.iloc[:,8]
model = XGBClassifier()
kfold = KFold(n_splits=10, random_state=7, shuffle=True)
results = cross_val_score(model, X,y, cv=kfold)
print("Mean Accuracy: %.2f%% (%.2f%%)" % (results.mean()*100, results.std()*100))
print("Median Accuracy: %.2f%% (%.2f%%)" % (np.median(results)*100, results.std()*100))
```

```python
df = pd.read_csv("pima-indians-diabetes.csv")
X = df.iloc[:,0:8]
y = df.iloc[:,8]
model = XGBClassifier()
kfold = StratifiedKFold(n_splits=10, random_state=7, shuffle=True)
results = cross_val_score(model, X,y, cv=kfold)
print("Mean Accuracy: %.2f%% (%.2f%%)" % (results.mean()*100, results.std()*100))
print("Median Accuracy: %.2f%% (%.2f%%)" % (np.median(results)*100, results.std()*100))
```

#### Visualize individual trees within a model

```python
from xgboost import to_graphviz
from IPython.display import display
from graphviz import Source

model = XGBClassifier()
model.fit(X, y)

graph = to_graphviz(model, tree_idx=0, rankdir="LR")

dot = graph.source.replace(
    "digraph {",
    'digraph {\n graph [ranksep=1.5, nodesep=0.8, dpi=150, size="30,40"];'
)
# Save as PNG
Source(dot).render("decision_tree_LR", format="png", cleanup=True)
# display(Source(dot))
```
